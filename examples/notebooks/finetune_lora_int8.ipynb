# xTuring task notebook: fine-tune with LoRA + INT8

Use this notebook for any supported model by changing `MODEL_KEY`.

```python
from pathlib import Path

from xturing.datasets.instruction_dataset import InstructionDataset
from xturing.models import BaseModel

DATASET_DIR = Path("../models/llama/alpaca_data")
MODEL_KEY = "qwen3_0_6b_lora_int8"
OUTPUT_DIR = Path("./outputs/lora_int8")
```

```python
dataset = InstructionDataset(str(DATASET_DIR))
model = BaseModel.create(MODEL_KEY)
model.finetune(dataset=dataset)
output = model.generate(texts=["How does quantization improve inference efficiency?"])
print(output)
model.save(str(OUTPUT_DIR))
```

See `https://xturing.stochastic.ai/overview/supported_models` for model keys and variant naming templates.
